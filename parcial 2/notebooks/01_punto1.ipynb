# Punto 1 (1.2) — Histograma y ecualización

Hemos mencionado dos funciones que permiten realizar el cómputo del histograma de una imagen. Estas funciones permiten identificar el número de pixeles que tienen cada una de las intensidades posibles para la imagen. Además, revisamos funciones que permiten realizar la ecualización de las imágenes. Completa la tabla con la información requerida y analiza los resultados.

\*\* Se recomienda el uso de funciones max y min e indexado de matrices (forma corta). Sin embargo, el ejercicio puede realizarse mediante ciclos (forma larga). Tenga en cuenta el valor inicial para los índices para el lenguaje que utilice para la solución.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
```

## Imagen en escala de grises



```python
img = np.array(Image.open("../images/mi_imagen.png").convert("RGB"))
gray = np.array(Image.fromarray(img).convert("L"))

plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.savefig("../results/punto1_gris.png", bbox_inches="tight", pad_inches=0)
```

## Histograma de la imagen en escala de grises



```python
hist = cv2.calcHist([gray], [0], None, [256], [0, 256]).flatten()

plt.figure()
plt.plot(hist)
plt.xlabel("Nivel de intensidad")
plt.ylabel("Numero de pixeles")
plt.savefig("../results/punto1_histograma_gris.png", bbox_inches="tight")
```

## Ecualización del histograma

Realice la ecualización del histograma sobre la imagen en escala de gris original.

```python
gray_eq = cv2.equalizeHist(gray)

plt.figure()
plt.imshow(gray_eq, cmap="gray")
plt.axis("off")
plt.savefig("../results/punto1_ecualizada.png", bbox_inches="tight", pad_inches=0)
```

## Histograma de la imagen ecualizada



```python
hist_eq = cv2.calcHist([gray_eq], [0], None, [256], [0, 256]).flatten()

plt.figure()
plt.plot(hist_eq)
plt.xlabel("Nivel de intensidad")
plt.ylabel("Numero de pixeles")
plt.savefig("../results/punto1_histograma_ecualizado.png", bbox_inches="tight")
```

## Nivel de intensidad más probable — imagen en gris

Identifique el nivel de intensidad uint8 de mayor probabilidad dentro de su imagen en escala de gris.

Nivel: **240**

Probabilidad: **4.44%**

```python
nivel_gris = int(np.argmax(hist))
prob_gris = hist[nivel_gris] / gray.size * 100
print(f"Nivel: {nivel_gris}")
print(f"Probabilidad: {prob_gris:.2f}%")
```

## Nivel de intensidad más probable — imagen ecualizada

Identifique el nivel de intensidad uint8 de mayor probabilidad dentro de su imagen ecualizada.

Nivel: **245**

Probabilidad: **4.44%**

```python
nivel_eq = int(np.argmax(hist_eq))
prob_eq = hist_eq[nivel_eq] / gray_eq.size * 100
print(f"Nivel: {nivel_eq}")
print(f"Probabilidad: {prob_eq:.2f}%")
```

## Probabilidad de intensidad = 176

Teniendo en cuenta la imagen ecualizada, ¿cuál es la probabilidad aproximada de que un píxel tenga una intensidad de 176?

Probabilidad: **0.87%**

```python
prob_176 = hist_eq[176] / gray_eq.size * 100
print(f"Probabilidad de intensidad 176: {prob_176:.4f}%")
```
