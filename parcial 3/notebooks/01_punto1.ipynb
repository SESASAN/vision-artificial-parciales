# Punto 1 (0.4) — Escala de grises y mapa de color

Convierta su imagen RGB a escala de grises con la función `rgb2gray`. Grafique su imagen usando el mapa de color el negativo del mapa de color parula.

Parula es un mapa de color de MATLAB y no existe en matplotlib: se usa su tabla de 64 colores (`PARULA` en `colorspaces.py`). El negativo del mapa de color se calcula como el complemento de cada color (1 − parula).

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
from skimage.color import rgb2gray
from matplotlib.colors import ListedColormap

import sys
sys.path.append("..")
from colorspaces import PARULA
```

## Imagen en escala de grises

```python
img = np.array(Image.open("../images/mi_imagen.png").convert("RGB"))
print(img.shape, img.dtype)

gray = np.round(rgb2gray(img) * 255).astype(np.uint8)

plt.figure()
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.savefig("../results/punto1_gris.png", bbox_inches="tight", pad_inches=0)
```

## Imagen graficada con el negativo de parula

Incorpore aquí la imagen resultante.

```python
parula_negativa = ListedColormap(1.0 - PARULA)

plt.figure()
plt.imshow(gray, cmap=parula_negativa, vmin=0, vmax=255)
plt.axis("off")
plt.savefig("../results/punto1_parula_negativa.png", bbox_inches="tight", pad_inches=0)
```

Identifique en su mapa de color la tripleta de valores RGB que se le asignan a una imagen de valores uint8 para el nivel de intensidad 170.
