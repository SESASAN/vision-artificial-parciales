# Punto 7 (0.5) — Realce de bordes Laplaciano de 4 vecinos

Aplique un filtro de realce de bordes laplaciano de 4 vecinos sobre su imagen RGB.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
```

## Imagen original RGB

```python
img = np.array(Image.open("../images/mi_imagen.png").convert("RGB"))
print(img.shape, img.dtype)

plt.figure()
plt.imshow(img)
plt.axis("off")
```

## Filtro de realce Laplaciano de 4 vecinos

El realce conserva la imagen y acentúa sus bordes: imagen original menos la respuesta del Laplaciano (el coeficiente central del kernel es negativo).

```python
laplaciano_4 = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]], dtype=np.float64)

respuesta = cv2.filter2D(img.astype(np.float64), -1, laplaciano_4, borderType=cv2.BORDER_REPLICATE)
resultado = np.clip(img - respuesta, 0, 255).astype(np.uint8)

Image.fromarray(resultado).save("../results/punto7_transformada.png")
```

## Comparación original vs filtrada

Complete la tabla con la imagen original RGB y la imagen filtrada.

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(resultado)
plt.axis("off")
plt.savefig("../results/punto7_comparacion.png", bbox_inches="tight", pad_inches=0)
```
